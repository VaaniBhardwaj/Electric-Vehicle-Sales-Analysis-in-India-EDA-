Cell 1: Importing Libraries and Loading Data
First, let's bring in the tools we need. We'll use pandas to work with our data table (the CSV file) and matplotlib to create our charts. Then, we'll load our "EV_Dataset.csv" file and do the same initial cleanup you did. This ensures our data is ready for analysis.

```python

import pandas as pd

import matplotlib.pyplot as plt
```

# Import the pandas library to read and work with our data
# Import the matplotlib library to create plots and graphs

# --- Load and Clean the Data ---
# Define the file path for our dataset

```python
file_path = "EV_Dataset.csv"
```

# Read the CSV file into a pandas DataFrame called 'df'

```python
df = pd.read_csv(file_path)
```

# --- Basic Data Cleaning ---

# Convert the 'EV_Sales_Quantity' column to numbers. If any value can't be converted,
# it becomes a 'NaN' (Not a Number), which we then fill with 0.

```python
df['EV_Sales_Quantity'] = pd.to_numeric(df['EV_Sales_Quantity'], errors='coerce').fillna(0)
```

# Clean up text columns by removing any extra spaces from the beginning or end

```python
df['State'] = df['State'].astype(str).str.strip()
df['Vehicle_Type'] = df['Vehicle_Type'].astype(str).str.strip()
df['Month_Name'] = df['Month_Name'].astype(str).str.strip()
```

# Convert the 'Year' column to whole numbers (integers)

```python
df['Year'] = pd.to_numeric(df['Year'], errors='coerce').fillna(0).astype(int)
```

# Let's look at the first 5 rows to make sure everything loaded correctly

```python
print(df.head())
```

Cell 2: Story - How Have EV Sales Grown Over the Years? 📈
Let's start with a big-picture view. Have electric vehicle sales been increasing over time? To find out, we can group all the sales data by year and then plot it. A line chart is perfect for showing a trend over time.

Cell 3: Code - Plotting Total EV Sales Per Year

# --- Yearly Sales Trend ---

# Group the data by 'Year' and calculate the total sum of 'EV_Sales_Quantity' for each year.

```python
yearly_sales = df.groupby('Year')['EV_Sales_Quantity'].sum()

# Let's see the numbers
print("Total EV Sales Per Year:")
print(yearly_sales)
```

# --- Create the Line Plot ---

# Set the size of our plot to be 10 inches wide by 6 inches tall

```python
plt.figure(figsize=(10, 6))

# Plot the yearly sales data. 'yearly_sales.index' is the Year, and 'yearly_sales.values' is the total sales.
plt.plot(yearly_sales.index, yearly_sales.values, marker='o', color='green')
```

```python
# Add a title to our chart
plt.title("Total EV Sales Growth Per Year")
# Add a label to the x-axis (the horizontal one)
plt.xlabel("Year")
# Add a label to the y-axis (the vertical one)
plt.ylabel("Total EV Sales Quantity")
# Add a grid to make the chart easier to read
plt.grid(True)
# Display the plot
plt.show()
```

Cell 4: Story - Which Vehicle Categories are the Most Popular? 🚗🛵
Now we know that sales are growing, but what kind of electric vehicles are people buying? Are they cars, scooters, or buses? A bar chart is a great way to compare the total sales of different categories side-by-side.

Cell 5: Code - Plotting Sales by Vehicle Category

# --- Sales by Vehicle Category ---

# Group the data by 'Vehicle_Category' and sum the sales for each one.

```python
category_sales = df.groupby('Vehicle_Category')['EV_Sales_Quantity'].sum()
```

# Sort the categories from most sales to least sales to make our chart clearer.

```python
category_sales = category_sales.sort_values(ascending=False)
```

# Let's see the numbers

```python
print("\nTotal EV Sales by Vehicle Category:")
print(category_sales)
```

# --- Create the Bar Chart ---

# Set the size of our plot

```python
plt.figure(figsize=(12, 7))
```

#Create a bar chart. 'category_sales.index' gives us the names of the categories,
# and 'category_sales.values' gives us the sales numbers.

```python
plt.bar(category_sales.index, category_sales.values, color='coral')
```

# Add a title, and labels for the x and y axes

```python
plt.title("Total EV Sales by Vehicle Category")
plt.xlabel("Vehicle Category")
plt.ylabel("Total EV Sales Quantity")
```

# Rotate the x-axis labels so they don't overlap

```python
plt.xticks(rotation=45)
```

```python
# Display the plot
plt.show()
```

Cell 6: Story - Which States are Leading the EV Charge? 🗺️
We know sales are growing and which vehicles are popular. But where is this happening? Let's find out which states have the highest number of EV sales. We will find the top 10 states and show them in a bar chart.

Cell 7: Code - Plotting Top 10 States by Sales

# --- Top 10 States by Sales ---

# Group the data by 'State' and sum the sales for each one.

```python
state_sales = df.groupby('State')['EV_Sales_Quantity'].sum()
```

# Sort the states by sales and keep only the top 10.

```python
top_10_states = state_sales.sort_values(ascending=False).head(10)
```

```python
# Let's see the numbers for the top 10
print("\nTop 10 States by EV Sales:")
print(top_10_states)
```

# --- Create the Bar Chart for Top 10 States ---

# Set the size of our plot

```python
plt.figure(figsize=(12, 7))
```

# Create a horizontal bar chart for better readability of state names.

```python
plt.barh(top_10_states.index, top_10_states.values, color='steelblue')
```

# Add a title and labels

```python
plt.title("Top 10 States by Total EV Sales")
plt.xlabel("Total EV Sales Quantity")
plt.ylabel("State")
# Invert y-axis to show the state with the highest sales at the top
plt.gca().invert_yaxis()
# Display the plot
plt.show()
```
